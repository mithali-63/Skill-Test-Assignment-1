```python
#Question1(A)
import numpy as np
import pandas as pd

# Sample data: 20 student scores
scores = [88, 76, 92, 67, 54, 45, 81, 95, 62, 73, 58, 49, 91, 85, 69, 77, 38, 84, 60, 66]

# Create a DataFrame
df = pd.DataFrame({'Student_ID': range(1, 21), 'Score': scores})

# 1. Overall Class Performance
average_score = np.mean(df['Score'])
highest_score = np.max(df['Score'])
lowest_score = np.min(df['Score'])
print("Class Performance:")
print(f"Average Score: {average_score:.2f}")
print(f"Highest Score: {highest_score}")
print(f"Lowest Score: {lowest_score}\n")

# 2. Pass/Fail Classification (Assuming pass mark is 50)
df['Result'] = df['Score'].apply(lambda x: 'Pass' if x >= 50 else 'Fail')
print("Pass/Fail Classification:")
print(df[['Student_ID', 'Score', 'Result']], "\n")

# 3. Grading on a Curve (Z-score based grading)
# Standardizing scores (Z-score normalization)
df['Z_score'] = (df['Score'] - df['Score'].mean()) / df['Score'].std()

# Assign grades based on Z-score thresholds (curved grading)
def assign_grade(z):
    if z >= 1.0:
        return 'A'
    elif z >= 0.5:
        return 'B'
    elif z >= 0.0:
        return 'C'
    elif z >= -0.5:
        return 'D'
    else:
        return 'F'

df['Grade'] = df['Z_score'].apply(assign_grade)

print("Graded Scores on a Curve:")
print(df[['Student_ID', 'Score', 'Z_score', 'Grade']])
```

```python
#Question 2(B)
import numpy as np
import matplotlib.pyplot as plt
# Each row = one student, each column = Math, Science, English
grades = np.random.randint(40, 101, (30, 3))
subjects = ['Math', 'Science', 'English']

# Calculate and print average, highest, and lowest for each subject
print("Subject-wise Statistics:")
for i in range(3):
    subject_grades = grades[:, i]
    print(f"{subjects[i]}:")
    print(f"  Average: {np.mean(subject_grades):.2f}")
    print(f"  Highest: {np.max(subject_grades)}")
    print(f"  Lowest: {np.min(subject_grades)}")

# Calculate total and average for each student
student_totals = np.sum(grades, axis=1)
student_averages = np.mean(grades, axis=1)
# Find top 3 students
top_students = np.argsort(student_totals)[-3:][::-1]
print("\nTop 3 Students:")
for idx in top_students:
    print(f"Student {idx + 1}: Total = {student_totals[idx]}, Average = {student_averages[idx]:.2f}")

# Optional: Show grade distributions with histograms
for i in range(3):
    plt.hist(grades[:, i], bins=10, color='lightblue', edgecolor='black')
    plt.title(f"{subjects[i]} Grade Distribution")
    plt.xlabel('Grades')
    plt.ylabel('Number of Students')
    plt.grid(True)
    plt.show()

```

```python
#Qusetion 3(A)
import numpy as np

# Frequency in Hz
frequency = 50

# Component values
R = 100         # Resistor in Ohms
L = 0.5         # Inductor in Henries
C = 100e-6      # Capacitor in Farads

# Choose configuration: "series" or "parallel"
configuration = "series"

# Calculate individual impedances
Z_R = R
Z_L = 1j * 2 * np.pi * frequency * L
Z_C = -1j / (2 * np.pi * frequency * C)

# Combine into a list
impedances = [Z_R, Z_L, Z_C]

# Calculate total impedance
if configuration == "series":
    Z_total = sum(impedances)
elif configuration == "parallel":
    Z_total = 1 / sum(1 / z for z in impedances)
else:
    print("Invalid configuration")
    Z_total = None

# Print the result
if Z_total is not None:
    print(f"Total impedance: {Z_total:.2f} ohms")

```

```python
#Question 3(B)
import numpy as np

# Step 1: Sample data for Stock A and Stock B (5 days, 3 features: Open, High, Low)
stock_A = np.array([ [100, 105, 98], [102, 108, 101], [104, 110, 103], [106, 112, 105], [108, 115, 107]])

stock_B = np.array([ [99, 104, 97],  [101, 107, 100],  [103, 109, 102],  [105, 111, 104],  [107, 114, 106]])

# Step 2: Element-wise sum
sum_matrix = np.add(stock_A, stock_B)

# Step 3: Element-wise difference
diff_matrix = np.subtract(stock_A, stock_B)

# Step 4: Output
print("=== Stock A Price Data ===")
print(stock_A)

print("\n=== Stock B Price Data ===")
print(stock_B)

print("\n=== Element-wise Sum (A + B) ===")
print(sum_matrix)

print("\n=== Element-wise Difference (A - B) ===")
print(diff_matrix)
```

```python
#Question 4(A)
import pandas as pd
import numpy as np

# Simulate EEG data: 3 participants, 4 electrodes, 100 time points
np.random.seed(42)  # For reproducibility
participants = ['P1', 'P2', 'P3']
electrodes = ['Fp1', 'Fp2', 'Cz', 'Pz']
time_points = 100

# Step 1: Create simulated data
data = []

for participant in participants:
    for electrode in electrodes:
        signal = np.random.normal(loc=0, scale=1, size=time_points)  # Simulated EEG signal
        for t, value in enumerate(signal):
            data.append([participant, electrode, t, value])

# Step 2: Load into DataFrame
df = pd.DataFrame(data, columns=['Participant', 'Electrode', 'Time', 'EEG_Value'])

# Step 3: Basic statistics per participant & electrode
stats = df.groupby(['Participant', 'Electrode'])['EEG_Value'].agg(['mean', 'var', 'max', 'min']).reset_index()

# Step 4: Display stats
print("=== Basic EEG Statistics per Participant & Electrode ===")
print(stats)

# Step 5: Optional - Identify patterns/trends using rolling average (example for P1, Fp1)
df_trend = df[(df['Participant'] == 'P1') & (df['Electrode'] == 'Fp1')].copy()
df_trend['Rolling_Mean'] = df_trend['EEG_Value'].rolling(window=5).mean()

# Optional: Visualization
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 4))
plt.plot(df_trend['Time'], df_trend['EEG_Value'], label='Raw EEG')
plt.plot(df_trend['Time'], df_trend['Rolling_Mean'], label='Rolling Mean (window=5)', color='red')
plt.title('EEG Pattern: Participant P1 - Electrode Fp1')
plt.xlabel('Time')
plt.ylabel('EEG Value')
plt.legend()
plt.tight_layout()
plt.show()




```

```python
#Question 4(B)
import pandas as pd
import matplotlib.pyplot as plt
data = { 'Product': ['Laptop', 'Smartphone', 'Tablet', 'Monitor', 'Keyboard'],'Price': [1200, 800, 300, 150, 50],'Quantity': [10, 20, 15, 25, 50]}
df = pd.DataFrame(data)

df['Total_Value'] = df['Price'] * df['Quantity']

total_inventory_value = df['Total_Value'].sum()

most_expensive = df.loc[df['Price'].idxmax()]
least_expensive = df.loc[df['Price'].idxmin()]
sorted_by_quantity = df.sort_values(by='Quantity', ascending=False)

filtered_products = df[df['Quantity'] > 15]
print("=== Inventory Data ===")
print(df)

print("\nTotal Inventory Worth: $", total_inventory_value)

print("\nMost Expensive Product:")
print(most_expensive)

print("\nLeast Expensive Product:")
print(least_expensive)

print("\nProducts Sorted by Quantity:")
print(sorted_by_quantity)

print("\nProducts with Quantity > 15:")
print(filtered_products)
plt.figure(figsize=(8, 4))
plt.bar(df['Product'], df['Quantity'], color='skyblue')
plt.title('Product Quantities in Inventory')
plt.xlabel('Product')
plt.ylabel('Quantity')
plt.grid(axis='y')
plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 6))
plt.pie(df['Total_Value'], labels=df['Product'], autopct='%1.1f%%', startangle=140)
plt.title('Inventory Value Distribution')
plt.tight_layout()
plt.show()
```

```python
#Question 5(A)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Generate Random Sales Data
np.random.seed(42)
months = pd.date_range(start='2024-01-01', periods=12, freq='MS')
sales = np.random.randint(10000, 50000, size=12)

# Create DataFrame
sales_df = pd.DataFrame({
    'Month': months,
    'Sales': sales
})

# Step 2: Save DataFrame to CSV
sales_df.to_csv('sales_data.csv', index=False)

# Step 3: Read the CSV file
df = pd.read_csv('sales_data.csv')

# Step 4: Display first 5 rows
print("First 5 rows of sales data:")
print(df.head())

# Step 5: Check for missing values
print("\nMissing values:")
print(df.isnull().sum())

# Optional: Fill or Drop (if any missing data is found)
df.dropna(inplace=True)  # In this case, we assume clean data

# Step 6: Visualize monthly sales trend
plt.figure(figsize=(10, 6))
sns.lineplot(data=df, x='Month', y='Sales', marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales Amount')
plt.xticks(rotation=45)
plt.tight_layout()
plt.grid(True)
plt.show()
```

```python
#Question 5(B)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(101)
num_records = 100
names = [f"Patient_{i}" for i in range(1, num_records + 1)]
ages = np.random.randint(20, 80, size=num_records)
genders = np.random.choice(['Male', 'Female'], size=num_records)
diagnoses = np.random.choice(['Diabetes', 'Hypertension', 'Asthma', 'Cancer', 'Healthy'], size=num_records, p=[0.25, 0.25, 0.2, 0.1, 0.2])

# Create DataFrame
healthcare_df = pd.DataFrame({'Name': names, 'Age': ages,'Gender': genders,'Diagnosis': diagnoses})
healthcare_df.to_csv('healthcare_data.csv', index=False)

df = pd.read_csv('healthcare_data.csv')
print("Column Names and Data Types:")
print(df.dtypes)
average_age = df['Age'].mean()
print(f"\nAverage Age of Patients: {average_age:.2f} years")

diagnosis_counts = df['Diagnosis'].value_counts()
print("\nDiagnosis Distribution:")
print(diagnosis_counts)

#Visualize the diagnosis distribution
plt.figure(figsize=(8, 6))
sns.countplot(data=df, x='Diagnosis', order=diagnosis_counts.index, palette='viridis')
plt.title('Distribution of Diagnoses')
plt.xlabel('Diagnosis')
plt.ylabel('Number of Patients')
plt.xticks(rotation=45)
plt.tight_layout()
plt.grid(True, axis='y')
plt.show()
```
